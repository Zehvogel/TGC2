```python
cat_sums = {}
cat_ns = {}

for cat in signal_category + interesting_bkg:
    sums_1ab = []
    for o_name in oo_names["mlvec_clean_brems_reco_oo"]:
        sums_1ab.append(analysis.get_sum(f"{o_name}_weight_hel_nominal", int_lumi=1000., categories=[cat]))
    n_per_ab = analysis.get_sum("weight_hel_nominal", int_lumi=1000., categories=[cat])
    cat_sums[cat] = sums_1ab
    cat_ns[cat] = n_per_ab
```

```python
# nominal means
nominal_means = []
for i in range(len(pars)):
    mean = cat_sums[signal_category[0]][i]  + cat_sums[interesting_bkg[0]][i]  + cat_sums[interesting_bkg[1]][i]
    mean /= cat_ns[signal_category[0]] + cat_ns[interesting_bkg[0]] + cat_ns[interesting_bkg[1]]
    nominal_means.append(mean)

var = 0.01

var1_means = []
for i in range(len(pars)):
    mean = cat_sums[signal_category[0]][i]  + (1 + var) * cat_sums[interesting_bkg[0]][i]  + cat_sums[interesting_bkg[1]][i]
    mean /= cat_ns[signal_category[0]] + (1 + var) * cat_ns[interesting_bkg[0]] + cat_ns[interesting_bkg[1]]
    var1_means.append(mean)

var2_means = []
for i in range(len(pars)):
    mean = cat_sums[signal_category[0]][i]  + cat_sums[interesting_bkg[0]][i]  + (1 + var) * cat_sums[interesting_bkg[1]][i]
    mean /= cat_ns[signal_category[0]] + cat_ns[interesting_bkg[0]] + (1 + var) * cat_ns[interesting_bkg[1]]
    var2_means.append(mean)
```

```python
# relative changes
rel_changes_var1 = [(v - m) / m for v, m in zip(var1_means, nominal_means)]
rel_changes_var2 = [(v - m) / m for v, m in zip(var2_means, nominal_means)]

# print results
print("Parameter\tNominal Mean\tVar1 Mean\tRel Change Var1\tVar2 Mean\tRel Change Var2")
for i, par in enumerate(pars):
    print(f"{par}\t{nominal_means[i]:.6f}\t{var1_means[i]:.6f}\t{rel_changes_var1[i]:.2%}\t{var2_means[i]:.6f}\t{rel_changes_var2[i]:.2%}")
```
