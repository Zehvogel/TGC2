```python

configs = {
    "ILC": {
        "ws1_path": "workspaces/new/pol_constrained_2.5em3/mlvec_clean_brems_reco_new_share_low_ppol_2ab.root",
        "ws2_path": "workspaces/new/pol_constrained_2.5em3/mlvec_clean_brems_reco_cc10_new_share_low_ppol_2ab.root",
        "par_names": par_names_lc,
        "n_runs": n_runs_lc,
    },
    "LCF": {
        "ws1_path": "workspaces/new/pol_constrained_2.5em3/mlvec_clean_brems_reco_new_share_low_ppol_3ab.root",
        "ws2_path": "workspaces/new/pol_constrained_2.5em3/mlvec_clean_brems_reco_cc10_new_share_low_ppol_3ab.root",
        "par_names": par_names_lc,
        "n_runs": n_runs_lc,
    },
    "FCC": {
        "ws1_path": "workspaces/new/pol_constrained_2.5em3/mlvec_clean_brems_reco_no_pol_10.8ab.root",
        "ws2_path": "workspaces/new/pol_constrained_2.5em3/mlvec_clean_brems_reco_cc10_no_pol_10.8ab.root",
        "par_names": par_names_cc,
        "n_runs": n_runs_cc,
    },
}
```

```python
config = configs[config_name]
ws1_path = config["ws1_path"]
ws2_path = config["ws2_path"]
par_names = config["par_names"]
n_runs = config["n_runs"]
ws1_path = "workspaces/pol_constrained_2.5em3/av_nurec_nomb_mc_new_share_low_ppol_2ab.root"
ws2_path = "workspaces/pol_constrained_2.5em3/av_nurec_nomb_mc_cc10_new_share_low_ppol_2ab.root"
# ws1_path = "workspaces/new/pol_constrained_2.5em3/av_nurec_nomb_mc_no_pol_2ab.root"
# ws2_path = "workspaces/new/pol_constrained_2.5em3/av_nurec_nomb_mc_cc10_no_pol_2ab.root"
```

```python
w = ROOT.RooWorkspace("w")
```

```python
w.Import(f"{ws1_path}:w:sim_model", RenameAllNodes="ws1")
w.Import(f"{ws2_path}:w:sim_model", RenameAllNodes="ws2")
```

```python
w.Print()
```

```python
cat = ROOT.RooCategory("runs_ws", "Run configurations", {f"ws_{i}_run_{j}": n_runs*(i-1)+j for i in [1, 2] for j in range(n_runs)})
cat.Print("v")
```

```python
# pdf_map = {f"ws_{i}_run_{j}": w.pdf(f"fit_model_run{j}_with_constraints_ws{i}") for i in [1, 2] for j in range(n_runs)}
pdf_map = {f"ws_{i}_run_{j}": w.pdf(f"fit_model_run{j}_with_constraints_ws{i}") for i in [1] for j in range(n_runs)}
pdf_map |= {f"ws_{i}_run_{j}": w.pdf(f"fit_model_run{j}_ws{i}") for i in [2] for j in range(n_runs)}
print(pdf_map)
```

```python
sim_model = ROOT.RooSimultaneous("sim_model", "Simultaneous fit model", pdf_map, cat)
w.Import(sim_model)
```

```python
obs_names = [f"O_{par}" for par in ["g1z", "ka", "la", "mW", "epol", "ppol"]]
obs = [w.var(name) for name in obs_names]
print(obs)
```

```python
ds = ROOT.RooStats.AsymptoticCalculator.GenerateAsimovData(sim_model, obs)
ds.printContents()
```

```python
# w.var("g1z").setConstant(True)
# w.var("ka").setConstant(True)
# w.var("la").setConstant(True)
w.var("mW").setConstant(True)
w.var("e_pol_L").setConstant(True)
w.var("e_pol_R").setConstant(True)
w.var("p_pol_L").setConstant(True)
w.var("p_pol_R").setConstant(True)

# w.var("e_pol_0").setConstant(True)
# w.var("p_pol_0").setConstant(True)
```

```python
nll = sim_model.createNLL(ds)
```

```python
nll_minimizer = ROOT.RooMinimizer(nll)
```

```python
nll_minimizer.migrad()
```

```python
nll_minimizer.hesse()
```

```python
res = nll_minimizer.lastMinuitFit()
res.Print()
```

```python
final_pars = res.floatParsFinal()
text = ""
for par_name in par_names:
    par = final_pars.find(par_name)
    if not par:
        continue
    text += f"{par_name}: {par.getVal()} ± {par.getError():.2e}\n"
print(text)
# with open(f"{plot_dir}/fit_results.txt", "w") as f:
#     f.write(text)
```

```python
name_dict = {
    "g1z": "g_{1}^{Z}",
    "ka": "#kappa_{#gamma}",
    "la": "#lambda_{#gamma}",
    "mW": "m_{W}",
    "e_pol_L": "P_{e^{-}}^{-}",
    "e_pol_R": "P_{e^{-}}^{+}",
    "e_pol_0": "P_{e^{-}}^{0}",
    "p_pol_L": "P_{e^{+}}^{-}",
    "p_pol_R": "P_{e^{+}}^{+}",
    "p_pol_0": "P_{e^{+}}^{0}",
}
```

```python
h = make_corr_mat(res, par_names, w, name_dict)
ch = ROOT.TCanvas()
h.SetMinimum(-1)
h.SetMaximum(1)
h.SetTitle(config_name)
h.Draw("col text")
h.SetMarkerSize(1.8)
ch.SetLeftMargin(0.1)
ch.SetRightMargin(0.05)
ch.SetBottomMargin(0.1)
ch.Draw()
ch.SaveAs(f"{plot_dir}/correlation_matrix.pdf")
```
