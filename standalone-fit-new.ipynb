```python
n_per_ab, nominal_means, slopes, Cov_O, means_list = load_fit_input(input_path)
```

```python
run_ab = 2
# run_ab = 11.2
```

```python
coupling_names = ["g1z", "ka", "lz"]
```

```python
Cov_mean_O = np.asarray(Cov_O) / (n_per_ab * run_ab)
```

```python
# expected precision:
np.sqrt(np.diag(np.linalg.inv(np.asarray(Cov_O))) / (n_per_ab * run_ab))
```
