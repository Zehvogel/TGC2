```python
workspaces = {}
datasets = {}

for name, input_path in input_paths.items():
    pars, n_per_ab, nominal_means, slopes, Cov_O, means_list = load_fit_input(input_path)
    Cov_mean_O = np.asarray(Cov_O) / (n_per_ab * run_ab)
    # expected precision:
    np.sqrt(np.diag(np.linalg.inv(np.asarray(Cov_O))) / (n_per_ab * run_ab))
    w = ROOT.RooWorkspace(f"{name}_w", "workspace")
    create_models(w, pars, nominal_means, slopes, Cov_mean_O)
    workspaces[name] = w
    w_datasets = make_datasets(w, "weight_hel", pars, means_list, x_points)
    datasets[name] = w_datasets
```

```python
fit_results = {}

for name, w in workspaces.items():
    w_datasets = datasets[name]
    w_fit_res = []
    for ds in w_datasets:
        # fit_res = fit(w, "fit_model", ds, minos=True)
        fit_res = fit(w, "fit_model", ds, minos=False, constant_parameters=constant_parameters)
        w_fit_res.append(fit_res)
    fit_results[name] = w_fit_res
```

```python
# config_name = "mlvec_clean_reco"
# config_name = "clean_brems_reco"
# config_name = "mlvec_clean_brems_reco"
config_name = "mlvec_clean_brems_reco_cc10"
# config_name = "av_nomb_mc"
# config_name = "av_nurec_nomb_mc_cc10"
# config_name = "kinfit_clean_reco_jm"
fit_res = fit_results[config_name]

nom = fit_res[0]
nom.correlationMatrix().Print()
nom.covarianceMatrix().Print()

c = ROOT.TCanvas()
h = nom.correlationHist()
h.Draw()
c.Draw("text colz0")
```

```python

graphs = {}

for meas_par in pars:
    if meas_par in constant_parameters:
        continue
    for i, var_par in enumerate(pars):
        x_vals = x_points.copy()
        y_vals = []
        y_errors_lo = []
        y_errors_hi = []
        n_x = len(x_vals)
        for j in range(n_x):
            f_res = fit_res[1 + i*n_x + j]
            res_par = f_res.floatParsFinal().find(meas_par)
            y_vals.append(res_par.getValV())
            y_errors_lo.append(res_par.getErrorLo())
            y_errors_hi.append(res_par.getErrorHi())
        x_vals.append(0.)
        nom_fit_par = fit_res[0].floatParsFinal().find(meas_par)
        y_vals.append(nom_fit_par.getValV())
        y_errors_lo.append(nom_fit_par.getErrorLo())
        y_errors_hi.append(nom_fit_par.getErrorHi())
        vals = sorted(zip(x_vals, y_vals, y_errors_lo, y_errors_hi), key=lambda a: a[0])
        print(vals)
        g = ROOT.TGraphAsymmErrors()
        for v in vals:
            g.AddPointError(v[0], v[1], 0., 0., -v[2], v[3])
        g.SetTitle(f";{var_par};{meas_par}")
        graphs[f"{meas_par}_{var_par}"] = g
```

```python
#  for par in pars:
# get all graphs where par is the varied parameters
# put them into a multigraph and plot them
multi_canvs = {}
legends = {}
multi_graphs = {}

plot_pars = ["g1z", "ka", "la", "mW"]
# plot_pars = pars

for var_par in pars:
    mg = ROOT.TMultiGraph()
    multi_graphs[var_par] = mg
    mg.SetTitle(f";{var_par};fitted value")
    legend = ROOT.TLegend(0.85, 0.70, 1., 1.)
    top_g = None
    for i, meas_par in enumerate(plot_pars):
        g = graphs[f"{meas_par}_{var_par}"]
        g.SetMarkerColor(ROOT.kP10Blue + i)
        g.SetLineColor(ROOT.kP10Blue + i)
        if meas_par == var_par:
            top_g = g
        else:
            # g.SetFillColorAlpha(ROOT.kP10Blue + i, 0.2)
            # mg.Add(g, "lp3")
            mg.Add(g, "lp")
        legend.AddEntry(g, meas_par, "lpf")
    if top_g is not None:
        mg.Add(top_g, "lp")
    c = ROOT.TCanvas()
    multi_canvs[var_par] = c
    legends[var_par] = legend
    mg.Draw("ALP")
    mg.GetXaxis().SetLimits(-0.0025, 0.0025)
    mg.SetMinimum(-0.0025)
    mg.SetMaximum(0.0025)
    legend.Draw()
    bottom_margin = 0.17
    top_margin = 0.08
    right_margin = 0.16
    left_margin = 0.15
    c.SetBottomMargin(bottom_margin)
    c.SetTopMargin(top_margin)
    c.SetRightMargin(right_margin)
    c.SetLeftMargin(left_margin)
    c.Draw()
    c.SaveAs(f"{plot_dir}/{config_name}_fitted_vs_{var_par}.pdf")
```
