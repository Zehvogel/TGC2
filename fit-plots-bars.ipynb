```python
def get_fit_results(fitres_dir: str, run_conf_names: list[str], lumi_list, oo_names):
    fit_results = {}
    for oo_name in oo_names:
        fs = {}
        for name in run_conf_names:
            for run_ab in lumi_list:
                run_name = f"{name}_{run_ab}ab"
                with ROOT.TFile.Open(f"{fitres_dir}/{oo_name}_{run_name}_fitres.root") as in_file:
                    fit_res = in_file.Get("fit_result_nominal")
                fs[run_name] = fit_res
        fit_results[oo_name] = fs
    return fit_results
```
