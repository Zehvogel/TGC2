```python
import ROOT
from math import sqrt
```

```python
oo_name = "av_nurec_nomb_mc"
run_name = "no_pol_3ab"
workspace_dir = "workspaces/pol_constrained_2.5em3"
```

```python
with ROOT.TFile(f"{workspace_dir}/{oo_name}_{run_name}.root") as f:
    w = f["w"]
```

```python
model = w.pdf("sim_model")
```

```python
w.loadSnapshot("nominal_parameters")
```

```python
asimov_ds = ROOT.RooStats.AsymptoticCalculator.GenerateAsimovData(model, w.set("observables"))
```

```python
nll = model.createNLL(asimov_ds)
```

```python
nll_minimizer = ROOT.RooMinimizer(nll)
nll_minimizer.migrad()
```

```python
nll_minimizer.hesse()
hesse_res = nll_minimizer.lastMinuitFit()
cov = hesse_res.covarianceMatrix()
cov.Print()
```

```python
poi_names = ["g1z", "ka", "la"]
poi = [w.var(poi_name) for poi_name in poi_names]
```

```python
profilell = nll.createProfile(poi)
```

```python
poi_set = ROOT.RooArgSet(poi)
li = ROOT.RooStats.LikelihoodInterval("li", profilell, poi_set)
```

```python
# li.SetConfidenceLevel(0.68)
li.SetConfidenceLevel(0.393)
```

```python
x_vals = ROOT.std.vector("double")(30, 0.)
y_vals = ROOT.std.vector("double")(30, 0.)
```

```python
li.GetContourPoints(poi[0], poi[1], x_vals.data(), y_vals.data(), 30)
```

```python
print(x_vals)
print(y_vals)
```

```python
profilell.getVariables().Print()
```

```python
c = ROOT.TCanvas()
g = ROOT.TGraph(30, x_vals.data(), y_vals.data())
# close the contour
g.SetPoint(30, x_vals[0], y_vals[0])
g.Draw("alp")
e = ROOT.RooEllipse("", 0., 0., sqrt(cov(1, 1)), sqrt(cov(2, 2)), cov(1, 2)/sqrt(cov(1, 1)*cov(2, 2)))
e.SetLineColor(ROOT.kRed)
e.Draw("l same")
c.Draw()
```
